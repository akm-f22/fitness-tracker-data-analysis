Duplicate value handling

```python
daily = daily.drop_duplicates()
intensity = intensity.drop_duplicates()
sleep = sleep.drop_duplicates()
weight = weight.drop_duplicates()
```

Converting and standardizing the date columns

```python
daily['ActivityDate'] = pd.to_datetime(daily['ActivityDate'], format='%m/%d/%Y %I:%M:%S %p')  # Added time component to format
intensity['ActivityDay'] = pd.to_datetime(intensity['ActivityDay'], format='%m/%d/%Y %I:%M:%S %p')  # Added time component to format
sleep['SleepDay'] = pd.to_datetime(sleep['SleepDay'], format='%m/%d/%Y %I:%M:%S %p')  # Added time component to format
weight['Date'] = pd.to_datetime(weight['Date'], format='%m/%d/%Y %I:%M:%S %p')  # Added time component to format
```

```python
sleep['Date'] = sleep['SleepDay'].dt.date
daily['Date'] = daily['ActivityDate'].dt.date
intensity['Date'] = intensity['ActivityDay'].dt.date
weight['Date'] = weight['Date'].dt.date
```

```python
sleep['Date'] = pd.to_datetime(sleep['Date'])
daily['Date'] = pd.to_datetime(daily['Date'])
intensity['Date'] = pd.to_datetime(intensity['Date'])
weight['Date'] = pd.to_datetime(weight['Date'])
```

Creating new features/columns

```python
daily['DayOfWeek'] = daily['Date'].dt.day_name()
daily['Month'] = daily['Date'].dt.month
sleep['SleepEfficiency'] = sleep['TotalMinutesAsleep'] / sleep['TotalTimeInBed']
```

Merging the datasets using Id and date

```python
merged = pd.merge(daily, intensity, on=['Id','Date'], how='left')

merged = pd.merge(merged,
                  sleep[['Id','Date','TotalMinutesAsleep','TotalTimeInBed','SleepEfficiency']],
                  on=['Id','Date'],
                  how='left')

merged = pd.merge(merged,
                  weight[['Id','Date','WeightKg','BMI']],
                  on=['Id','Date'],
                  how='left')
```

Exporting final dataset

```python
merged.isnull().sum()
```

```python
merged['TotalMinutesAsleep'] = merged['TotalMinutesAsleep'].ffill()
merged['TotalTimeInBed'] = merged['TotalTimeInBed'].ffill()
merged['SleepEfficiency'] = merged['SleepEfficiency'].ffill()
merged['WeightKg'] = merged['WeightKg'].ffill()
merged['BMI'] = merged['BMI'].ffill()
```
